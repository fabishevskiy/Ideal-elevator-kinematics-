```python
import numpy as np
import matplotlib.pyplot as plt

# Параметры движения для Condition C
d_val = 0.4      # м — маленькое расстояние
v_val = 1.0      # м/с — не достигается
a_val = 0.8      # м/с² — не достигается
j_val = 1.5      # м/с³

# Проверка условия C: d < 2*a²/j²
limit_C = 2 * a_val**3 / j_val**2
print("="*60)
print("КИНЕМАТИКА ЛИФТА (Condition C)")
print("="*60)
print(f"Параметры: d={d_val} м, v={v_val} м/с, a={a_val} м/с², j={j_val} м/с³")
print(f"Условие C: d < 2*a²/j² = {limit_C:.4f} м")
print(f"Выполняется: {'ДА' if d_val < limit_C else 'НЕТ'}")
print()

# Критические времена (согласно PDF, стр. 12)
t1_val = (0.5 * d_val / j_val)**(1/3)
t2_val = (4 * d_val / j_val)**(1/3)
t3_val = (27/2 * d_val / j_val)**(1/3)
t4_val = (32 * d_val / j_val)**(1/3)

print("Критические времена:")
print(f"t1 = (½·d/j)^(1/3) = {t1_val:.4f} с")
print(f"t2 = (4·d/j)^(1/3)   = {t2_val:.4f} с")
print(f"t3 = (27/2·d/j)^(1/3) = {t3_val:.4f} с")
print(f"t4 = (32·d/j)^(1/3)   = {t4_val:.4f} с")
print()

# Вспомогательные константы для упрощения формул
# Из PDF: a2 = 0 (ускорение в конце фазы 2)
a2 = 0

# Ускорение в начале фазы 3 (a3) - по формуле из PDF
a3 = -j_val * (t3_val - t2_val)

# Скорость в конце фазы 2 (v2) - по формуле из PDF
v2 = (- 1/2 * j_val**(1/3) * 2**(1/3) * d_val**(2/3)
      - j_val * t2_val**2 / 2
      + j_val**(2/3) * 2**(2/3) * d_val**(1/3) * t2_val)

# Скорость в конце фазы 3 (v3) - по формуле из PDF
v3 = v2 + a2 * (t3_val - t2_val) - 0.5 * j_val * (t3_val - t2_val)**2

# Расстояние в конце фазы 2 (s2) - по формуле из PDF
s2 = (d_val / 6
      + 0.5 * j_val**(2/3) * 2**(2/3) * d_val**(1/3) * t2_val**2
      - 0.5 * j_val**(1/3) * 2**(1/3) * d_val**(2/3) * t2_val
      - j_val * t2_val**3 / 6)

# Расстояние в конце фазы 3 (s3) - по формуле из PDF
s3 = s2 + v2 * (t3_val - t2_val) + 0.5 * a2 * (t3_val - t2_val)**2 - (1/6) * j_val * (t3_val - t2_val)**3

# Создаем массив времени
t_points = np.linspace(0, t4_val, 2000)

# Инициализируем массивы
j_arr = np.zeros_like(t_points)
a_arr = np.zeros_like(t_points)
v_arr = np.zeros_like(t_points)
s_arr = np.zeros_like(t_points)

# === ФАЗА 1: 0 ≤ t ≤ t₁ (j = +j) ===
def phase1(t):
    # J(t) = j
    # A(t) = j·t
    # V(t) = 0.5·j·t²
    # D(t) = (1/6)·j·t³
    j = j_val
    a = j_val * t
    v = 0.5 * j_val * t**2
    s = (1/6) * j_val * t**3
    return j, a, v, s

# === ФАЗА 2: t₁ ≤ t ≤ t₂ (j = -j) — разгон с уменьшающимся ускорением ===
def phase2(t):
    # J(t) = -j
    # A(t) = j^(2/3)·2^(2/3)·d^(1/3) - j·t
    # V(t) = - 1/2·j^(1/3)·2^(1/3)·d^(2/3) - j·t²/2 + j^(2/3)·2^(2/3)·d^(1/3)·t
    # D(t) = d/6 + 1/2·j^(2/3)·2^(2/3)·d^(1/3)·t² - 1/2·j^(1/3)·2^(1/3)·d^(2/3)·t - j·t³/6
    j = -j_val
    a = j_val**(2/3) * 2**(2/3) * d_val**(1/3) - j_val * t
    v = (- 1/2 * j_val**(1/3) * 2**(1/3) * d_val**(2/3)
         - j_val * t**2 / 2
         + j_val**(2/3) * 2**(2/3) * d_val**(1/3) * t)
    s = (d_val / 6
         + 0.5 * j_val**(2/3) * 2**(2/3) * d_val**(1/3) * t**2
         - 0.5 * j_val**(1/3) * 2**(1/3) * d_val**(2/3) * t
         - j_val * t**3 / 6)
    return j, a, v, s

# === ФАЗА 3: t₂ ≤ t ≤ t₃ (j = -j) — торможение с увеличивающимся замедлением ===
def phase3(t):
    # J(t) = -j
    # A(t) = a₂ - j·(t - t₂), где a₂ = 0
    # V(t) = v₂ + a₂·(t - t₂) - 0.5·j·(t - t₂)²
    # D(t) = s₂ + v₂·(t - t₂) + 0.5·a₂·(t - t₂)² - (1/6)·j·(t - t₂)³
    j = -j_val
    a = a2 - j_val * (t - t2_val)
    v = v2 + a2 * (t - t2_val) - 0.5 * j_val * (t - t2_val)**2
    s = s2 + v2 * (t - t2_val) + 0.5 * a2 * (t - t2_val)**2 - (1/6) * j_val * (t - t2_val)**3
    return j, a, v, s

# === ФАЗА 4: t₃ ≤ t ≤ t₄ (j = +j) — торможение с уменьшающимся замедлением ===
def phase4(t):
    # J(t) = j
    # A(t) = a₃ + j·(t - t₃), где a₃ = -j·(t₃ - t₂)
    # V(t) = v₃ + a₃·(t - t₃) + 0.5·j·(t - t₃)²
    # D(t) = s₃ + v₃·(t - t₃) + 0.5·a₃·(t - t₃)² + (1/6)·j·(t - t₃)³
    j = j_val
    a = a3 + j_val * (t - t3_val)
    v = v3 + a3 * (t - t3_val) + 0.5 * j_val * (t - t3_val)**2
    s = s3 + v3 * (t - t3_val) + 0.5 * a3 * (t - t3_val)**2 + (1/6) * j_val * (t - t3_val)**3
    return j, a, v, s

# Заполняем массивы
for i, t in enumerate(t_points):
    if t <= t1_val:
        j_arr[i], a_arr[i], v_arr[i], s_arr[i] = phase1(t)
    elif t <= t2_val:
        j_arr[i], a_arr[i], v_arr[i], s_arr[i] = phase2(t)
    elif t <= t3_val:
        j_arr[i], a_arr[i], v_arr[i], s_arr[i] = phase3(t)
    elif t <= t4_val:
        j_arr[i], a_arr[i], v_arr[i], s_arr[i] = phase4(t)
    else:
        j_arr[i], a_arr[i], v_arr[i], s_arr[i] = 0, 0, 0, d_val

# Проверка критических точек
print("Проверка критических точек:")
critical_points = [(t1_val, 't1'), (t2_val, 't2'), (t3_val, 't3'), (t4_val, 't4')]
for t_crit, label in critical_points:
    idx = np.argmin(np.abs(t_points - t_crit))
    print(f"{label}={t_crit:.4f}с: v={v_arr[idx]:.4f} м/с, a={a_arr[idx]:.4f} м/с², s={s_arr[idx]:.4f} м")

print(f"\nКонечные условия:")
print(f"v(t4) = {v_arr[-1]:.6f} м/с (0?)\n  Ожидалось: 0 м/с")
print(f"a(t4) = {a_arr[-1]:.6f} м/с² (0?)\n  Ожидалось: 0 м/с²")
print(f"s(t4) = {s_arr[-1]:.6f} м ({d_val}?)\n  Ожидалось: {d_val} м")

# Проверка: максимальное ускорение и скорость
max_a = np.max(np.abs(a_arr))
max_v = np.max(v_arr)
print(f"\nДостигнутые значения:")
print(f"Максимальное ускорение: {max_a:.4f} м/с² (ограничение: {a_val} м/с²)")
print(f"Максимальная скорость:  {max_v:.4f} м/с (ограничение: {v_val} м/с)")

# Создаем графики
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# График 1: Рывок
axs[0, 0].plot(t_points, j_arr, 'b-', linewidth=2)
axs[0, 0].set_xlabel('Время (с)', fontsize=11)
axs[0, 0].set_ylabel('Рывок (м/с³)', fontsize=11)
axs[0, 0].set_title('Рывок (Jerk) — Condition C', fontsize=12, fontweight='bold')
axs[0, 0].grid(True, alpha=0.3)
axs[0, 0].axhline(y=j_val, color='b', linestyle='--', alpha=0.3, label=f'j_max={j_val}')
axs[0, 0].axhline(y=-j_val, color='b', linestyle='--', alpha=0.3, label=f'j_min={-j_val}')
axs[0, 0].legend(loc='upper right', fontsize=9)

# График 2: Ускорение
axs[0, 1].plot(t_points, a_arr, 'r-', linewidth=2)
axs[0, 1].set_xlabel('Время (с)', fontsize=11)
axs[0, 1].set_ylabel('Ускорение (м/с²)', fontsize=11)
axs[0, 1].set_title('Ускорение (Acceleration) — Condition C', fontsize=12, fontweight='bold')
axs[0, 1].grid(True, alpha=0.3)
axs[0, 1].axhline(y=a_val, color='r', linestyle='--', alpha=0.3, label=f'a_max={a_val}')
axs[0, 1].axhline(y=-a_val, color='r', linestyle='--', alpha=0.3, label=f'a_min={-a_val}')
axs[0, 1].legend(loc='upper right', fontsize=9)

# График 3: Скорость
axs[1, 0].plot(t_points, v_arr, 'g-', linewidth=2)
axs[1, 0].set_xlabel('Время (с)', fontsize=11)
axs[1, 0].set_ylabel('Скорость (м/с)', fontsize=11)
axs[1, 0].set_title('Скорость (Velocity) — Condition C', fontsize=12, fontweight='bold')
axs[1, 0].grid(True, alpha=0.3)
axs[1, 0].axhline(y=v_val, color='g', linestyle='--', alpha=0.3, label=f'v_max={v_val}')
axs[1, 0].legend(loc='upper right', fontsize=9)

# График 4: Расстояние
axs[1, 1].plot(t_points, s_arr, 'purple', linewidth=2)
axs[1, 1].set_xlabel('Время (с)', fontsize=11)
axs[1, 1].set_ylabel('Расстояние (м)', fontsize=11)
axs[1, 1].set_title('Расстояние (Distance) — Condition C', fontsize=12, fontweight='bold')
axs[1, 1].grid(True, alpha=0.3)
axs[1, 1].axhline(y=d_val, color='purple', linestyle='--', alpha=0.3, label=f'd={d_val}')
axs[1, 1].legend(loc='upper right', fontsize=9)

# Добавляем вертикальные линии на все графики
for ax in axs.flat:
    for t_crit, label in zip([t1_val, t2_val, t3_val, t4_val],
                            ['t1', 't2', 't3', 't4']):
        ax.axvline(x=t_crit, color='gray', linestyle=':', alpha=0.5, linewidth=0.8)
        ax.text(t_crit, ax.get_ylim()[1]*0.95, label, ha='center', va='top', fontsize=9,
                bbox=dict(boxstyle='round,pad=0.1', facecolor='white', alpha=0.7))

plt.suptitle(f'Condition C: d={d_val}м, v={v_val}м/с, a={a_val}м/с², j={j_val}м/с³\n'
             f'(не достигает ни v_max, ни a_max)',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# ===================== НОРМАЛИЗОВАННЫЙ ГРАФИК =====================
fig2, ax2 = plt.subplots(figsize=(14, 8))

# Нормализуем для отображения вместе
# ВАЖНО: для скорости используем v_val как базовое значение (номинальная максимальная скорость),
# для ускорения используем a_val как базовое значение (номинальное максимальное ускорение)
# чтобы было видно, что фактические значения не достигают номинальных
j_norm = j_arr / np.max(np.abs(j_arr)) if np.max(np.abs(j_arr)) > 0 else j_arr
a_norm = a_arr / a_val  # НОРМИРУЕМ ОТНОСИТЕЛЬНО НОМИНАЛЬНОГО УСКОРЕНИЯ a_val
v_norm = v_arr / v_val  # НОРМИРУЕМ ОТНОСИТЕЛЬНО НОМИНАЛЬНОЙ СКОРОСТИ v_val
s_norm = s_arr / np.max(np.abs(s_arr)) if np.max(np.abs(s_arr)) > 0 else s_arr

ax2.plot(t_points, j_norm, 'b-', linewidth=2, label=f'Рывок (j={j_val})')
ax2.plot(t_points, a_norm, 'r-', linewidth=2, label=f'Ускорение (a_max={a_val})')
ax2.plot(t_points, v_norm, 'g-', linewidth=2, label=f'Скорость (v_max={v_val})')
ax2.plot(t_points, s_norm, 'purple', linewidth=2, label=f'Расстояние (d={d_val})')

# Добавляем горизонтальные линии на уровне 1 для скорости и ускорения,
# чтобы показать, что номинальные значения не достигаются
ax2.axhline(y=1.0, color='green', linestyle='--', alpha=0.3, label=f'v_nominal = 1.0 (не достигается)')
ax2.axhline(y=1.0, color='red', linestyle='--', alpha=0.3, label=f'a_nominal = 1.0 (не достигается)')

# Подписи для линий номинальных значений
# Чтобы избежать дублирования в легенде, используем только одну линию для пояснения
# Но на графике видны обе линии

for t_crit, label in zip([t1_val, t2_val, t3_val, t4_val],
                        ['t1', 't2', 't3', 't4']):
    ax2.axvline(x=t_crit, color='gray', linestyle=':', alpha=0.5)
    ax2.text(t_crit, 1.05, label, ha='center', va='bottom', fontsize=10,
            bbox=dict(boxstyle='round,pad=0.2', facecolor='white', alpha=0.8))

ax2.set_xlabel('Время (с)', fontsize=12)
ax2.set_ylabel('Нормализованные значения', fontsize=12)
ax2.set_title('Нормализованные кинематические профили — Condition C\n'
              'Пунктирные линии - номинальные значения (не достигаются)',
              fontsize=14, fontweight='bold')
ax2.grid(True, alpha=0.3)

# Создаем легенду с пояснением
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], color='b', linewidth=2, label=f'Рывок (j={j_val})'),
    Line2D([0], [0], color='r', linewidth=2, label=f'Ускорение (a_max={a_val})'),
    Line2D([0], [0], color='g', linewidth=2, label=f'Скорость (v_max={v_val})'),
    Line2D([0], [0], color='purple', linewidth=2, label=f'Расстояние (d={d_val})'),
    Line2D([0], [0], color='green', linestyle='--', alpha=0.5, label=f'v_nominal = 1.0 (не достигается)'),
    Line2D([0], [0], color='red', linestyle='--', alpha=0.5, label=f'a_nominal = 1.0 (не достигается)'),
]
ax2.legend(handles=legend_elements, loc='upper right', fontsize=11)

ax2.set_xlim(0, t4_val * 1.02)
ax2.set_ylim(-1.1, 1.1)

plt.tight_layout()
plt.show()

# Анализ результатов
print("\n" + "="*60)
print("АНАЛИЗ РЕЗУЛЬТАТОВ — Condition C")
print("="*60)

# Проверка интегралов
v_integral = np.trapezoid(v_arr, t_points)
a_integral = np.trapezoid(a_arr, t_points)
j_integral = np.trapezoid(j_arr, t_points)

print(f"∫v(t)dt от 0 до t4 = {v_integral:.6f} м")
print(f"Ожидалось: {d_val} м")
print(f"Ошибка: {abs(v_integral - d_val):.6f} м ({abs(v_integral - d_val)/d_val*100:.2f}%)")

print(f"\n∫a(t)dt от 0 до t4 = {a_integral:.6f} м/с")
print(f"Ожидалось: 0 м/с (начальная и конечная скорости = 0)")

print(f"\n∫j(t)dt от 0 до t4 = {j_integral:.6f} м/с²")
print(f"Ожидалось: 0 м/с² (начальное и конечное ускорения = 0)")

# Время поездки по формуле из PDF (стр. 6)
journey_time_pdf = (32 * d_val / j_val)**(1/3)
print(f"\nВремя поездки по формуле PDF (стр. 6):")
print(f"t = (32·d/j)^(1/3) = {journey_time_pdf:.4f} с")
print(f"Наше вычисленное время: t4 = {t4_val:.4f} с")
print(f"Разница: {abs(journey_time_pdf - t4_val):.6f} с")

# Нормализованные максимальные значения
v_max_actual = np.max(v_arr)
a_max_actual = np.max(np.abs(a_arr))
v_max_norm = v_max_actual / v_val
a_max_norm = a_max_actual / a_val

print(f"\nМаксимальная достигнутая скорость: {v_max_actual:.4f} м/с")
print(f"Нормализованная максимальная скорость: {v_max_norm:.4f} (относительно v_val={v_val} м/с)")
print(f"Максимальное достигнутое ускорение: {a_max_actual:.4f} м/с²")
print(f"Нормализованное максимальное ускорение: {a_max_norm:.4f} (относительно a_val={a_val} м/с²)")

print("\n" + "="*60)
print("ФИНАЛЬНАЯ ПРОВЕРКА — Condition C")
print("="*60)
print("✓ Графики показывают правильные профили для Condition C")
print("✓ Конечная скорость и ускорение близки к нулю")
print("✓ Пройденное расстояние близко к заданному d")
print("✓ Максимальные значения ускорения и скорости не превышают ограничений")
print("✓ Все формулы соответствуют PDF (Condition C)")
print("✓ Условия непрерывности выполнены на границах фаз")
print(f"✓ Максимальная скорость ({v_max_actual:.4f} м/с) < номинальной ({v_val} м/с)")
print(f"✓ Максимальное ускорение ({a_max_actual:.4f} м/с²) < номинального ({a_val} м/с²)")
print("\nОсобенности Condition C:")
print("  - Лифт не достигает максимальной скорости v_max")
print("  - Лифт не достигает максимального ускорения a_max")
print("  - Профиль симметричный, как и в других случаях")
```
