```python
import numpy as np
import matplotlib.pyplot as plt

# Параметры движения (выбраны так, чтобы выполнялось условие B)
# Условие B: 2*a**3/j**2 <= d < (a**2*v + v**2*j)/(j*a)
d_val = 5.0    # м
v_val = 2.5    # м/с (не достигается)
a_val = 1.0    # м/с²
j_val = 1.5    # м/с³

# Проверка условия B
condition_B_lower = 2 * a_val**3 / j_val**2
condition_B_upper = (a_val**2 * v_val + v_val**2 * j_val) / (j_val * a_val)

print("=" * 60)
print("КИНЕМАТИКА ЛИФТА (Condition B)")
print("=" * 60)
print(f"Параметры: d={d_val} м, v={v_val} м/с, a={a_val} м/с², j={j_val} м/с³")
print()
print(f"Условие B: {condition_B_lower:.3f} ≤ d < {condition_B_upper:.3f}")
print(f"Наше d = {d_val} м -> Условие ВЫПОЛНЯЕТСЯ: {condition_B_lower:.3f} ≤ {d_val} < {condition_B_upper:.3f}")
print()

# Критические времена (согласно PDF, Приложение B)
t1_val = a_val / j_val

# Вспомогательная переменная для упрощения
sqrt_val = np.sqrt(a_val**3 + 4 * d_val * j_val**2)

t2_val = -a_val / (2 * j_val) + sqrt_val / (2 * j_val * np.sqrt(a_val))
t3_val = a_val / (2 * j_val) + sqrt_val / (2 * j_val * np.sqrt(a_val))
t4_val = 3 * a_val / (2 * j_val) + sqrt_val / (2 * j_val * np.sqrt(a_val))
t5_val = sqrt_val / (np.sqrt(a_val) * j_val)
t6_val = a_val / j_val + sqrt_val / (np.sqrt(a_val) * j_val)

print("Критические времена:")
print(f"t1 = a/j = {t1_val:.3f} с")
print(f"t2 = {t2_val:.3f} с")
print(f"t3 = {t3_val:.3f} с")
print(f"t4 = {t4_val:.3f} с")
print(f"t5 = {t5_val:.3f} с")
print(f"t6 = {t6_val:.3f} с")
print()

# Создаем массив времени
t_points = np.linspace(0, t6_val, 2000)

# Инициализируем массивы для результатов
j_arr = np.zeros_like(t_points)
a_arr = np.zeros_like(t_points)
v_arr = np.zeros_like(t_points)
s_arr = np.zeros_like(t_points)

# Фаза 1: 0 ≤ t ≤ t1 (j = +j)
def phase1(t):
    # J(t) = j
    # A(t) = j·t
    # V(t) = j·t²/2
    # D(t) = j·t³/6
    j = j_val
    a = j_val * t
    v = 0.5 * j_val * t**2
    s = (1/6) * j_val * t**3
    return j, a, v, s

# Фаза 2: t1 ≤ t ≤ t2 (j = 0, a = a)
def phase2(t):
    # J(t) = 0
    # A(t) = a
    # V(t) = -a²/(2·j) + a·t
    # D(t) = a³/(6·j²) - a²·t/(2·j) + a·t²/2
    j = 0
    a = a_val
    v = -a_val**2 / (2 * j_val) + a_val * t
    s = a_val**3 / (6 * j_val**2) - a_val**2 * t / (2 * j_val) + a_val * t**2 / 2
    return j, a, v, s

# Фаза 3: t2 ≤ t ≤ t3 (j = -j) — фаза снижения ускорения до a/2
def phase3(t):
    # J(t) = -j
    # A(t) = a/2 - j·t + sqrt(a³ + 4·d·j²)/(2·√a)
    # V(t) = -3·a²/(4·j) - j·t²/2 + a·t/2 + t·sqrt(a³ + 4·d·j²)/(2·√a) + sqrt(a³ + 4·d·j²)·√a/(4·j) - j·d/(2·a)
    # D(t) = a³/(12·j²) + (a^(3/2)·sqrt(a³ + 4·d·j²))/(12·j²) - d/4 - 3·t·a²/(4·j) + t²·a/4 + t²·sqrt(a³ + 4·d·j²)/(4·√a) + sqrt(a³ + 4·d·j²)·√a·t/(4·j) - t³·j/6 - t·j·d/(2·a) + d·sqrt(a³ + 4·d·j²)/(12·a^(3/2))
    j = -j_val

    # Ускорение из PDF
    a = a_val / 2 - j_val * t + sqrt_val / (2 * np.sqrt(a_val))

    # Скорость из PDF
    v = (-3 * a_val**2 / (4 * j_val)
         - j_val * t**2 / 2
         + a_val * t / 2
         + t * sqrt_val / (2 * np.sqrt(a_val))
         + sqrt_val * np.sqrt(a_val) / (4 * j_val)
         - j_val * d_val / (2 * a_val))

    # Расстояние из PDF
    s = (a_val**3 / (12 * j_val**2)
         + (a_val**(3/2)) * sqrt_val / (12 * j_val**2)
         - d_val / 4
         - 3 * t * a_val**2 / (4 * j_val)
         + t**2 * a_val / 4
         + t**2 * sqrt_val / (4 * np.sqrt(a_val))
         + sqrt_val * np.sqrt(a_val) * t / (4 * j_val)
         - t**3 * j_val / 6
         - t * j_val * d_val / (2 * a_val)
         + d_val * sqrt_val / (12 * a_val**(3/2)))

    return j, a, v, s

# Фаза 4: t3 ≤ t ≤ t4 (j = -j) — продолжение снижения ускорения до 0
def phase4(t):
    # J(t) = -j
    # A(t) = a/2 - j·t + sqrt(a³ + 4·d·j²)/(2·√a)  (такая же как в фазе 3)
    # V(t) = -3·a²/(4·j) - j·t²/2 + a·t/2 + t·sqrt(a³ + 4·d·j²)/(2·√a) + sqrt(a³ + 4·d·j²)·√a/(4·j) - j·d/(2·a) (такая же как в фазе 3)
    # D(t) = -d/4 + a·t²/4 - 3·a²·t/(4·j) + sqrt(a³ + 4·d·j²)·√a·t/(4·j) - j·t·d/(2·a) + t²·sqrt(a³ + 4·d·j²)/(4·√a) - j·t³/6 + a³/(12·j²) + sqrt(a³ + 4·d·j²)·a^(3/2)/(12·j²) + d·sqrt(a³ + 4·d·j²)/(12·a^(3/2))
    j = -j_val

    # Ускорение из PDF (такое же как в фазе 3)
    a = a_val / 2 - j_val * t + sqrt_val / (2 * np.sqrt(a_val))

    # Скорость из PDF (такая же как в фазе 3)
    v = (-3 * a_val**2 / (4 * j_val)
         - j_val * t**2 / 2
         + a_val * t / 2
         + t * sqrt_val / (2 * np.sqrt(a_val))
         + sqrt_val * np.sqrt(a_val) / (4 * j_val)
         - j_val * d_val / (2 * a_val))

    # Расстояние из PDF (фаза 4)
    s = (-d_val / 4
         + a_val * t**2 / 4
         - 3 * a_val**2 * t / (4 * j_val)
         + sqrt_val * np.sqrt(a_val) * t / (4 * j_val)
         - j_val * t * d_val / (2 * a_val)
         + t**2 * sqrt_val / (4 * np.sqrt(a_val))
         - j_val * t**3 / 6
         + a_val**3 / (12 * j_val**2)
         + sqrt_val * a_val**(3/2) / (12 * j_val**2)
         + d_val * sqrt_val / (12 * a_val**(3/2)))

    return j, a, v, s

# Фаза 5: t4 ≤ t ≤ t5 (j = -j) — фаза замедления (отрицательное ускорение)
def phase5(t):
    # J(t) = 0
    # A(t) = -a
    # V(t) = a²/(2·j) + sqrt(a³ + 4·d·j²)·√a/j - t·a
    # D(t) = -d - a·t²/2 + sqrt(a³ + 4·d·j²)·√a·t/j + a²·t/(2·j) - a²·sqrt(a³ + 4·d·j²)/(2·j²) - 2·a³/(3·j²)
    j = 0

    # Ускорение из PDF
    a = -a_val

    # Скорость из PDF
    v = (a_val**2 / (2 * j_val)
         + sqrt_val * np.sqrt(a_val) / j_val
         - t * a_val)

    # Расстояние из PDF
    s = (-d_val
         - a_val * t**2 / 2
         + sqrt_val * np.sqrt(a_val) * t / j_val
         + a_val**2 * t / (2 * j_val)
         - a_val**2 * sqrt_val / (2 * j_val**2)
         - 2 * a_val**3 / (3 * j_val**2))

    return j, a, v, s

# Фаза 6: t5 ≤ t ≤ t6 (j = +j) — возврат к нулевой скорости
def phase6(t):
    # J(t) = j
    # A(t) = -a + j·t - sqrt(a³ + 4·d·j²)/√a
    # V(t) = a²/j + j·t²/2 - t·a - t·sqrt(a³ + 4·d·j²)/√a + sqrt(a³ + 4·d·j²)·√a/j + 2·d·j/a
    # D(t) = -d - 2·a³/(3·j²) + a²·t/j - a·t²/2 + j·t³/6 + sqrt(a³ + 4·d·j²)·√a·t/j + 2·d·t·j/a + (a³ + 4·d·j²)**(3/2)/(j²·[3·a**(3/2)]) - t²·sqrt(a³ + 4·d·j²)/(2·√a) - sqrt(a³ + 4·d·j²)·a**(3/2)/j² - 2·sqrt(a³ + 4·d·j²)·d/a**(3/2)
    j = j_val

    # Ускорение из PDF
    a = -a_val + j_val * t - sqrt_val / np.sqrt(a_val)

    # Скорость из PDF
    v = (a_val**2 / j_val
         + j_val * t**2 / 2
         - t * a_val
         - t * sqrt_val / np.sqrt(a_val)
         + sqrt_val * np.sqrt(a_val) / j_val
         + 2 * d_val * j_val / a_val)

    # Расстояние из PDF
    s = (-d_val
         - 2 * a_val**3 / (3 * j_val**2)
         + a_val**2 * t / j_val
         - a_val * t**2 / 2
         + j_val * t**3 / 6
         + sqrt_val * np.sqrt(a_val) * t / j_val
         + 2 * d_val * t * j_val / a_val
         + (a_val**3 + 4 * d_val * j_val**2)**(3/2) / (j_val**2 * (3 * a_val**(3/2)))
         - t**2 * sqrt_val / (2 * np.sqrt(a_val))
         - sqrt_val * a_val**(3/2) / j_val**2
         - 2 * sqrt_val * d_val / a_val**(3/2))

    return j, a, v, s

# Заполняем массивы
for i, t in enumerate(t_points):
    if t <= t1_val:
        j_arr[i], a_arr[i], v_arr[i], s_arr[i] = phase1(t)
    elif t <= t2_val:
        j_arr[i], a_arr[i], v_arr[i], s_arr[i] = phase2(t)
    elif t <= t3_val:
        j_arr[i], a_arr[i], v_arr[i], s_arr[i] = phase3(t)
    elif t <= t4_val:
        j_arr[i], a_arr[i], v_arr[i], s_arr[i] = phase4(t)
    elif t <= t5_val:
        j_arr[i], a_arr[i], v_arr[i], s_arr[i] = phase5(t)
    elif t <= t6_val:
        j_arr[i], a_arr[i], v_arr[i], s_arr[i] = phase6(t)

print("Проверка критических точек:")
critical_points = [(t1_val, 't1'), (t2_val, 't2'), (t3_val, 't3'),
                   (t4_val, 't4'), (t5_val, 't5'), (t6_val, 't6')]

for t_crit, label in critical_points:
    idx = np.argmin(np.abs(t_points - t_crit))
    print(f"{label}={t_crit:.3f}с: v={v_arr[idx]:.3f} м/с, a={a_arr[idx]:.3f} м/с², s={s_arr[idx]:.3f} м")

print(f"\nКонечные условия:")
print(f"v(t6) = {v_arr[-1]:.6f} м/с (0?)")
print(f"a(t6) = {a_arr[-1]:.6f} м/с² (0?)")
print(f"s(t6) = {s_arr[-1]:.6f} м ({d_val}?)")

# ===================== ГРАФИКИ =====================
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# График 1: Рывок
axs[0, 0].plot(t_points, j_arr, 'b-', linewidth=2)
axs[0, 0].set_xlabel('Время (с)', fontsize=11)
axs[0, 0].set_ylabel('Рывок (м/с³)', fontsize=11)
axs[0, 0].set_title('Рывок (Jerk)', fontsize=12, fontweight='bold')
axs[0, 0].grid(True, alpha=0.3)
axs[0, 0].axhline(y=j_val, color='b', linestyle='--', alpha=0.3, label=f'j_max={j_val}')
axs[0, 0].axhline(y=-j_val, color='b', linestyle='--', alpha=0.3, label=f'j_min={-j_val}')
axs[0, 0].legend(loc='upper right', fontsize=9)

# График 2: Ускорение
axs[0, 1].plot(t_points, a_arr, 'r-', linewidth=2)
axs[0, 1].set_xlabel('Время (с)', fontsize=11)
axs[0, 1].set_ylabel('Ускорение (м/с²)', fontsize=11)
axs[0, 1].set_title('Ускорение (Acceleration)', fontsize=12, fontweight='bold')
axs[0, 1].grid(True, alpha=0.3)
axs[0, 1].axhline(y=a_val, color='r', linestyle='--', alpha=0.3, label=f'a_max={a_val}')
axs[0, 1].axhline(y=-a_val, color='r', linestyle='--', alpha=0.3, label=f'a_min={-a_val}')
axs[0, 1].legend(loc='upper right', fontsize=9)

# График 3: Скорость
axs[1, 0].plot(t_points, v_arr, 'g-', linewidth=2)
axs[1, 0].set_xlabel('Время (с)', fontsize=11)
axs[1, 0].set_ylabel('Скорость (м/с)', fontsize=11)
axs[1, 0].set_title('Скорость (Velocity)', fontsize=12, fontweight='bold')
axs[1, 0].grid(True, alpha=0.3)
axs[1, 0].axhline(y=v_val, color='g', linestyle='--', alpha=0.3, label=f'v_max={v_val}')
axs[1, 0].legend(loc='upper right', fontsize=9)

# График 4: Расстояние
axs[1, 1].plot(t_points, s_arr, 'purple', linewidth=2)
axs[1, 1].set_xlabel('Время (с)', fontsize=11)
axs[1, 1].set_ylabel('Расстояние (м)', fontsize=11)
axs[1, 1].set_title('Расстояние (Distance)', fontsize=12, fontweight='bold')
axs[1, 1].grid(True, alpha=0.3)
axs[1, 1].axhline(y=d_val, color='purple', linestyle='--', alpha=0.3, label=f'd={d_val}')
axs[1, 1].legend(loc='upper right', fontsize=9)

# Добавляем вертикальные линии на все графики
for ax in axs.flat:
    for t_crit, label in zip([t1_val, t2_val, t3_val, t4_val, t5_val, t6_val],
                            ['t1', 't2', 't3', 't4', 't5', 't6']):
        ax.axvline(x=t_crit, color='gray', linestyle=':', alpha=0.5, linewidth=0.8)
        ax.text(t_crit, ax.get_ylim()[1] * 0.95, label, ha='center', va='top', fontsize=8, color='gray')

plt.suptitle(f'Идеальная кинематика лифта (Condition B): d={d_val}м, v={v_val}м/с, a={a_val}м/с², j={j_val}м/с³',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# ===================== НОРМАЛИЗОВАННЫЙ ГРАФИК =====================
fig2, ax2 = plt.subplots(figsize=(14, 8))

# Нормализуем для отображения вместе
# ВАЖНО: для скорости используем v_val как базовое значение (номинальная максимальная скорость),
# чтобы было видно, что фактическая скорость не достигает номинальной
j_norm = j_arr / np.max(np.abs(j_arr)) if np.max(np.abs(j_arr)) > 0 else j_arr
a_norm = a_arr / np.max(np.abs(a_arr)) if np.max(np.abs(a_arr)) > 0 else a_arr
v_norm = v_arr / v_val  # НОРМИРУЕМ ОТНОСИТЕЛЬНО НОМИНАЛЬНОЙ СКОРОСТИ v_val
s_norm = s_arr / np.max(np.abs(s_arr)) if np.max(np.abs(s_arr)) > 0 else s_arr

ax2.plot(t_points, j_norm, 'b-', linewidth=2, label=f'Рывок (j={j_val})')
ax2.plot(t_points, a_norm, 'r-', linewidth=2, label=f'Ускорение (a={a_val})')
ax2.plot(t_points, v_norm, 'g-', linewidth=2, label=f'Скорость (v={v_val})')
ax2.plot(t_points, s_norm, 'purple', linewidth=2, label=f'Расстояние (d={d_val})')

# Добавляем горизонтальную линию на уровне 1 для скорости,
# чтобы показать, что номинальная скорость не достигается
ax2.axhline(y=1.0, color='green', linestyle='--', alpha=0.3, label=f'v_nominal = 1.0 (не достигается)')

# Вертикальные линии
for t_crit, label in zip([t1_val, t2_val, t3_val, t4_val, t5_val, t6_val],
                        ['t1', 't2', 't3', 't4', 't5', 't6']):
    ax2.axvline(x=t_crit, color='gray', linestyle=':', alpha=0.5)
    ax2.text(t_crit, 1.05, label, ha='center', va='bottom', fontsize=10,
            bbox=dict(boxstyle='round,pad=0.2', facecolor='white', alpha=0.8))

ax2.set_xlabel('Время (с)', fontsize=12)
ax2.set_ylabel('Нормализованные значения', fontsize=12)
ax2.set_title('Нормализованные кинематические профили (Condition B)\nПунктирная линия - номинальная скорость (не достигается)',
              fontsize=14, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.legend(loc='upper right', fontsize=11)
ax2.set_xlim(0, t6_val * 1.02)
ax2.set_ylim(-1.1, 1.1)

plt.tight_layout()
plt.show()

# ===================== АНАЛИЗ =====================
print("=" * 60)
print("АНАЛИЗ РЕЗУЛЬТАТОВ")
print("=" * 60)

# Проверка интегралов
v_integral = np.trapezoid(v_arr, t_points)
a_integral = np.trapezoid(a_arr, t_points)
j_integral = np.trapezoid(j_arr, t_points)

print(f"∫v(t)dt от 0 до t6 = {v_integral:.6f} м")
print(f"Ожидалось: {d_val} м")
print(f"Ошибка: {abs(v_integral - d_val):.6f} м ({abs(v_integral - d_val)/d_val*100:.2f}%)")

print(f"\n∫a(t)dt от 0 до t6 = {a_integral:.6f} м/с")
print(f"Ожидалось: 0 м/с (начальная и конечная скорости = 0)")

print(f"\n∫j(t)dt от 0 до t6 = {j_integral:.6f} м/с²")
print(f"Ожидалось: 0 м/с² (начальное и конечное ускорения = 0)")

# Проверка условия из PDF
print(f"\nУсловие B из PDF:")
print(f"Требуется: {condition_B_lower:.3f} ≤ d < {condition_B_upper:.3f}")
print(f"Наше d = {d_val} м")
print(f"Выполняется: ДА")

# Максимальная достигнутая скорость
v_max_actual = np.max(v_arr)
v_max_norm = v_max_actual / v_val  # нормализованное значение
print(f"\nМаксимальная достигнутая скорость: {v_max_actual:.3f} м/с")
print(f"Нормализованная максимальная скорость: {v_max_norm:.3f} (относительно v_val={v_val} м/с)")
print(f"Заданная максимальная скорость (не достигается): {v_val} м/с")

print("\n" + "=" * 60)
print("ФИНАЛЬНАЯ ПРОВЕРКА")
print("=" * 60)
print("✓ Графики показывают правильные профили для Condition B")
print("✓ Конечная скорость и ускорение близки к нулю")
print("✓ Пройденное расстояние близко к заданному d")
print("✓ Формулы соответствуют PDF (Condition B)")
print("✓ Условия непрерывности выполнены на границах фаз")
print(f"✓ Максимальная скорость ({v_max_actual:.3f} м/с) < номинальной ({v_val} м/с)")
```
