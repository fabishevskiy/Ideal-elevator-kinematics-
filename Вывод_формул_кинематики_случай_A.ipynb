```python
# @title Импорт библиотек
import sympy as sp
from sympy import symbols, integrate, simplify, pprint, init_printing, solve, Eq, S

# Для красивого отображения формул
init_printing(use_unicode=True)

print("Библиотеки успешно загружены!")
```

```python
# @title Определяем символы для времени и параметров движения

t, j, a, v, d = symbols('t j a v d', real=True, positive=True)
tau, sigma = symbols('tau sigma', real=True, positive=True)

print("\nОпределены символы:")
print(f"t - время")
print(f"j - рывок (максимальный jerk) в м/с³")
print(f"a - ускорение (максимальное) в м/с²")
print(f"v - скорость (максимальная) в м/с")
print(f"d - расстояние (путь) в м")
print(f"tau, sigma - вспомогательные переменные для интегрирования")

print("\nСимволы для символьных вычислений:")
print(f"t = {t}")
print(f"j = {j}")
print(f"a = {a}")
print(f"v = {v}")
print(f"d = {d}")
```

```python
# @title Этап 1: Интервал 0 ≤ t ≤ t1 (разгон с постоянным jerk = j)
# Запустите эту ячейку после Этапа 1

print("=== ИНТЕРВАЛ 1: 0 ≤ t ≤ t1 ===\n")

# В этом интервале:
# J(t) = j (постоянный рывок)
# A(t) = j*t (ускорение растет от 0 до a)
# V(t) = j*t²/2 (скорость растет от 0)
# D(t) = j*t³/6 (расстояние растет от 0)

# Записываем функции
J1 = j
print("J(t):")
pprint(J1)
print()

A1 = integrate(J1, t)
print("A(t) = ∫J(t)dt =")
pprint(A1)
print()

V1 = integrate(A1, t)
print("V(t) = ∫A(t)dt =")
pprint(V1)
print()

D1 = integrate(V1, t)
print("D(t) = ∫V(t)dt =")
pprint(D1)
print()

# Проверка: в момент t1 ускорение должно быть равно a
t1 = a/j
print(f"В момент t1 = a/j = {t1}:")
print(f"J(t1) = {J1}")
print(f"A(t1) = {A1.subs(t, t1)} = a")
print(f"V(t1) = {V1.subs(t, t1)} = a²/(2j)")
print(f"D(t1) = {D1.subs(t, t1)} = a³/(6j²)")

print("\n" + "="*60 + "\n")

# Сохраняем значения в конце интервала для использования в следующем интервале
J1_end = J1
A1_end = A1.subs(t, t1)
V1_end = V1.subs(t, t1)
D1_end = D1.subs(t, t1)

print("Значения в конце интервала (t = t1):")
print(f"J(t1) = {J1_end}")
print(f"A(t1) = {A1_end}")
print(f"V(t1) = {V1_end}")
print(f"D(t1) = {D1_end}")
```

```python
# @title Этап 2: Вывод t2 и t3
# Запустите эту ячейку после Этапа 2

print("=== ВЫВОД t2 И t3 ИЗ ФИЗИЧЕСКИХ УСЛОВИЙ ===\n")

# Из первого интервала мы знаем:
t1 = a/j
V1_end = j*t1**2 / 2  # V(t1) = a²/(2j)
D1_end = j*t1**3 / 6  # D(t1) = a³/(6j²)

print(f"Из интервала 1:")
print(f"t1 = {t1}")
print(f"V(t1) = {V1_end}")
print(f"D(t1) = {D1_end}\n")

# Выводим t2 из условия, что в момент t3 скорость достигает v
# В интервале t1→t2: V(t) = V(t1) + a*(t - t1)
# В интервале t2→t3: A(t) = a - j*(t - t2), V(t3) = V(t2) + a²/(2j)
# V(t3) = V(t1) + a*(t2 - t1) + a²/(2j) = v

print("Вывод t2 из условия V(t3) = v:")
print("V(t3) = V(t1) + a*(t2 - t1) + a²/(2j) = v")
print(f"a²/(2j) + a*(t2 - a/j) + a²/(2j) = v")
print(f"a*t2 - a²/j + a²/j = v")
print(f"a*t2 = v")
print(f"t2 = v/a")

# Вычисляем t2
t2 = v/a
print(f"\nt2 = {t2}")

# Вычисляем t3
t3 = t2 + a/j
print(f"t3 = t2 + a/j = {t3}")
```

```python
# @title Этап 3: Интервал t1 ≤ t ≤ t2
print("=== ИНТЕРВАЛ 2: t1 ≤ t ≤ t2 ===\n")
print("(постоянное ускорение a, jerk = 0)")

print(f"Границы интервала:")
print(f"t1 = {t1}")
print(f"t2 = {t2}\n")

# Значения в конце первого интервала
V1_end = j*t1**2 / 2
D1_end = j*t1**3 / 6

# В этом интервале:
# J(t) = 0
# A(t) = a
# V(t) = V(t1) + a*(t - t1)
# D(t) = D(t1) + V(t1)*(t - t1) + a*(t - t1)²/2

J2 = S(0)
print("J(t):")
pprint(J2)
print()

A2 = a
print("A(t):")
pprint(A2)
print()

V2 = V1_end + integrate(A2, (t, t1, t))
V2 = simplify(V2)
print("V(t) = V(t1) + ∫A(t)dt =")
pprint(V2)
print()

D2 = D1_end + integrate(V2, (t, t1, t))
D2 = simplify(D2)
print("D(t) = D(t1) + ∫V(t)dt =")
pprint(D2)
print()

# Проверка в момент t2
V2_end = V2.subs(t, t2)
D2_end = D2.subs(t, t2)
print(f"Проверка в момент t2 = {t2}:")
print(f"V(t2) = {V2_end}")
print(f"V(t2) должно быть = v - a²/(2j)")
print(f"Проверка: {simplify(V2_end - (v - a**2/(2*j)))} = 0\n")

print("Значения в конце интервала (t = t2):")
print(f"J(t2) = {J2}")
print(f"A(t2) = {A2}")
print(f"V(t2) = {V2_end}")
print(f"D(t2) = {D2_end}")

print("\n" + "="*60 + "\n")
```

```python
# @title Этап 4: Интервал t2 ≤ t ≤ t3 (ускорение уменьшается от a до 0, jerk = -j)
# Запустите эту ячейку после Этапа 3

print("=== ИНТЕРВАЛ 3: t2 ≤ t ≤ t3 ===\n")
print("(ускорение уменьшается от a до 0, jerk = -j)")

# Границы интервала
t1 = a/j
t2 = v/a
t3 = v/a + a/j

print(f"Границы интервала:")
print(f"t2 = {t2}")
print(f"t3 = {t3}\n")

# Значения в начале интервала (из предыдущего)
V2_end = v - a**2/(2*j)
D2_end = simplify((a**3)/(6*j**2) + (a*v**2)/(2*j**2) + (a**2*v)/(2*j**2) + (v**2)/(2*a) - (a*v)/(2*j))

print(f"V(t2) = {V2_end}")
print(f"D(t2) = {D2_end}\n")

# В этом интервале:
# J(t) = -j
# A(t) = a - j*(t - t2) (ускорение падает от a до 0)
# V(t) = V(t2) + ∫A(t)dt
# D(t) = D(t2) + ∫V(t)dt

J3 = -j
print("J(t):")
pprint(J3)
print()

# A(t) = a + ∫(-j)dt от t2 до t
A3 = a + integrate(J3, (t, t2, t))
A3 = simplify(A3)
print("A(t) = a + ∫J(t)dt =")
pprint(A3)
print()

V3 = V2_end + integrate(A3, (t, t2, t))
V3 = simplify(V3)
print("V(t) = V(t2) + ∫A(t)dt =")
pprint(V3)
print()

D3 = D2_end + integrate(V3, (t, t2, t))
D3 = simplify(D3)
print("D(t) = D(t2) + ∫V(t)dt =")
pprint(D3)
print()

# Проверка в момент t3
A3_end = A3.subs(t, t3)
V3_end = V3.subs(t, t3)
D3_end = D3.subs(t, t3)

print(f"Проверка в момент t3 = {t3}:")
print(f"A(t3) = {A3_end} = 0")
print(f"Проверка: {simplify(A3_end)} = 0")
print(f"V(t3) = {V3_end} = v")
print(f"Проверка: {simplify(V3_end - v)} = 0\n")

print("Значения в конце интервала (t = t3):")
print(f"J(t3) = {J3}")
print(f"A(t3) = {A3_end}")
print(f"V(t3) = {V3_end}")
print(f"D(t3) = {D3_end}")

print("\n=== ИТОГ ДЛЯ ПЕРВЫХ ТРЕХ ИНТЕРВАЛОВ ===")
print("Мы прошли три интервала разгона:")
print("1. 0→t1: разгон с постоянным jerk = j (ускорение от 0 до a)")
print("2. t1→t2: постоянное ускорение a (jerk = 0)")
print("3. t2→t3: замедление набора ускорения (jerk = -j, ускорение от a до 0)")
print(f"\nВ момент t3 = {t3} достигается максимальная скорость v = {v}")
print(f"Расстояние, пройденное к моменту t3: D(t3) = {D3_end}")
```

```python
# @title Этап 5: Вывод t4 (правильный)
# Запустите эту ячейку после Этапа 4

print("=== ВЫВОД t4 (ВРЕМЯ НАЧАЛА ТОРМОЖЕНИЯ) ===\n")

# Известные значения
t1 = a/j
t2 = v/a
t3 = v/a + a/j

print(f"t1 = {t1}")
print(f"t2 = {t2}")
print(f"t3 = {t3}\n")

# Вычисляем D(t3) - расстояние разгона
V1_end = j*t1**2 / 2
D1_end = j*t1**3 / 6

# Функции для интервалов разгона
V2 = V1_end + a*(t - t1)
D2 = D1_end + V1_end*(t - t1) + a*(t - t1)**2/2
V3 = V2.subs(t, t2) + a*(t - t2) - j*(t - t2)**2/2
D3 = D2.subs(t, t2) + V2.subs(t, t2)*(t - t2) + a*(t - t2)**2/2 - j*(t - t2)**3/6

# D(t3)
D_t3 = D3.subs(t, t3)
D_t3 = simplify(D_t3)
print(f"Расстояние разгона D(t3) = {D_t3}")

# Проверяем соотношение v*t3 = 2*D(t3)
v_t3 = v*t3
two_D_t3 = 2*D_t3
print(f"\nv*t3 = {v_t3}")
print(f"2*D(t3) = {two_D_t3}")
print(f"Разность: {simplify(v_t3 - two_D_t3)} = 0")
print(f"Проверка: v*t3 = 2*D(t3) ? {simplify(v_t3 - two_D_t3) == 0}")

print("\n" + "="*60 + "\n")

# Теперь выводим t4
print("Из симметрии профиля:")
print("d = D(t3) + v*(t4 - t3) + D(t3)")
print("d = 2*D(t3) + v*(t4 - t3)")
print("v*(t4 - t3) = d - 2*D(t3)")
print("t4 - t3 = (d - 2*D(t3))/v")
print("t4 = t3 + (d - 2*D(t3))/v")

# Подставляем 2*D(t3) = v*t3 (из проверки выше)
print("\nТак как 2*D(t3) = v*t3, то:")
print("t4 = t3 + (d - v*t3)/v")
print("t4 = t3 + d/v - t3")
print("t4 = d/v")

t4 = d/v
print(f"\nt4 = {t4}")

```

```python
# @title Этап 6 (исправленный): Проверка и вывод t6, t7

print("=== ПРОВЕРКА ЗНАЧЕНИЙ t6 И t7 ===\n")

# Известные значения
t1 = a/j
t2 = v/a
t3 = v/a + a/j
t4 = d/v
t5 = t4 + a/j

print(f"t1 = {t1}")
print(f"t2 = {t2}")
print(f"t3 = {t3}")
print(f"t4 = {t4}")
print(f"t5 = {t5}\n")

# Вычисляем V(t5)
V_t5 = v - a**2/(2*j)
print(f"V(t5) = {V_t5}\n")

# t6 = t5 + V(t5)/a
print("Из физических условий:")
print("t6 = t5 + V(t5)/a")
t6_calc = t5 + V_t5/a
t6_calc = simplify(t6_calc)
print(f"t6 = {t6_calc}")

# Проверяем, равно ли это d/v + v/a
t6_expected = d/v + v/a
print(f"\nОжидаемое значение из файла: t6 = d/v + v/a = {t6_expected}")

# Сравниваем
diff6 = simplify(t6_calc - t6_expected)
print(f"Разница: {diff6} = 0? {diff6 == 0}")

print("\n" + "="*60 + "\n")

# t7 = t6 + a/j
print("t7 = t6 + a/j")
t7_calc = t6_calc + a/j
t7_calc = simplify(t7_calc)
print(f"t7 = {t7_calc}")

# Проверяем, равно ли это d/v + a/j + v/a
t7_expected = d/v + a/j + v/a
print(f"\nОжидаемое значение из файла: t7 = d/v + a/j + v/a = {t7_expected}")

# Сравниваем
diff7 = simplify(t7_calc - t7_expected)
print(f"Разница: {diff7} = 0? {diff7 == 0}")

print("\n" + "="*60 + "\n")

print("=== ПРОВЕРКА СИММЕТРИИ ПРОФИЛЯ ===\n")

# Время разгона
T_accel = t3
print(f"Время разгона: T_accel = t3 = {t3}")

# Время торможения
T_brake = t7 - t4
print(f"Время торможения: T_brake = t7 - t4 = {T_brake}")

# Проверка симметрии
print(f"T_accel = T_brake? {simplify(T_accel - T_brake) == 0}")

print("\nРасстояние разгона = расстоянию торможения?")
# Вычисляем D(t3)
V1_end = j*t1**2 / 2
D1_end = j*t1**3 / 6
V2 = V1_end + a*(t - t1)
D2 = D1_end + V1_end*(t - t1) + a*(t - t1)**2/2
V3 = V2.subs(t, t2) + a*(t - t2) - j*(t - t2)**2/2
D3 = D2.subs(t, t2) + V2.subs(t, t2)*(t - t2) + a*(t - t2)**2/2 - j*(t - t2)**3/6

D_t3 = D3.subs(t, t3)
D_t3 = simplify(D_t3)
print(f"D(t3) = {D_t3}")

# Расстояние торможения = d - D(t4) = d - d = 0? Нет, D(t4) = d
# Расстояние торможения = d - D(t3) - v*(t4 - t3)
D_brake = d - D_t3 - v*(t4 - t3)
D_brake = simplify(D_brake)
print(f"Расстояние торможения = {D_brake}")
print(f"D(t3) = D_brake? {simplify(D_t3 - D_brake) == 0}")

print("\n" + "="*60 + "\n")

print("=== ОКОНЧАТЕЛЬНЫЕ ЗНАЧЕНИЯ ВРЕМЕННЫХ ТОЧЕК ===\n")
print(f"t1 = {t1} (достижение ускорения a)")
print(f"t2 = {t2} (начало уменьшения ускорения)")
print(f"t3 = {t3} (достижение скорости v, ускорение=0)")
print(f"t4 = {t4} = d/v (начало торможения)")
print(f"t5 = {t5} (достижение ускорения -a)")
print(f"t6 = {t6_expected} = d/v + v/a (начало завершения торможения)")
print(f"t7 = {t7_expected} = d/v + a/j + v/a (конец движения)")

print("\nПроверка:")
print(f"t6 = t5 + V(t5)/a = {t6_calc}")
print(f"t6 (из файла) = {t6_expected}")
print(f"Совпадают? {simplify(t6_calc - t6_expected) == 0}")

print(f"\nt7 = t6 + a/j = {t7_calc}")
print(f"t7 (из файла) = {t7_expected}")
print(f"Совпадают? {simplify(t7_calc - t7_expected) == 0}")

print("\n=== ОБЩЕЕ ВРЕМЯ ДВИЖЕНИЯ ===")
print(f"T = t7 = {t7_expected}")
print(f"Формула из файла: T = d/v + a/j + v/a")
print(f"Проверка: {simplify(t7_expected - (d/v + a/j + v/a))} = 0")
```

```python
# @title Этап 6: Интервал t3 ≤ t ≤ t4 (движение с постоянной скоростью v)
# Запустите эту ячейку после Этапа 5

print("=== ИНТЕРВАЛ 4: t3 ≤ t ≤ t4 ===\n")
print("(движение с постоянной скоростью v)")

# Границы интервала
t1 = a/j
t2 = v/a
t3 = v/a + a/j
t4 = d/v

print(f"Границы интервала:")
print(f"t3 = {t3}")
print(f"t4 = {t4}\n")

# Вычисляем D(t3) - расстояние в конце разгона
V1_end = j*t1**2 / 2
D1_end = j*t1**3 / 6
V2 = V1_end + a*(t - t1)
D2 = D1_end + V1_end*(t - t1) + a*(t - t1)**2/2
V3 = V2.subs(t, t2) + a*(t - t2) - j*(t - t2)**2/2
D3 = D2.subs(t, t2) + V2.subs(t, t2)*(t - t2) + a*(t - t2)**2/2 - j*(t - t2)**3/6

D_t3 = D3.subs(t, t3)
D_t3 = simplify(D_t3)
print(f"D(t3) = {D_t3}\n")

# В этом интервале:
# J(t) = 0 (рынок равен нулю)
# A(t) = 0 (ускорение равно нулю)
# V(t) = v (скорость постоянная)
# D(t) = D(t3) + v*(t - t3)

J4 = S(0)
print("J(t):")
pprint(J4)
print()

A4 = S(0)
print("A(t):")
pprint(A4)
print()

V4 = v
print("V(t):")
pprint(V4)
print()

D4 = D_t3 + integrate(V4, (t, t3, t))
print("D(t) = D(t3) + ∫V(t)dt =")
pprint(D4)
print()

# Проверка в момент t4
D4_end = D4.subs(t, t4)
print(f"Проверка в момент t4 = {t4}:")
print(f"J(t4) = {J4}")
print(f"A(t4) = {A4}")
print(f"V(t4) = {V4} = v")
print(f"D(t4) = {D4_end} = d")
print(f"Проверка: {simplify(D4_end - d)} = 0")

print("\n" + "="*60 + "\n")

print("Значения в конце интервала (t = t4):")
print(f"J(t4) = {J4}")
print(f"A(t4) = {A4}")
print(f"V(t4) = {V4}")
print(f"D(t4) = {D4_end}")
```

```python
# @title Этап 7: Интервал t4 ≤ t ≤ t5 (торможение, ускорение от 0 до -a, jerk = -j)
# Запустите эту ячейку после Этапа 6

print("=== ИНТЕРВАЛ 5: t4 ≤ t ≤ t5 ===\n")
print("(торможение, ускорение от 0 до -a, jerk = -j)")

# Границы интервала
t1 = a/j
t2 = v/a
t3 = v/a + a/j
t4 = d/v
t5 = t4 + a/j
t6 = d/v + v/a
t7 = d/v + a/j + v/a

print(f"Границы интервала:")
print(f"t4 = {t4}")
print(f"t5 = {t5}\n")

# В этом интервале:
# J(t) = -j (постоянный отрицательный рывок)
# A(t) = -j*(t - t4) (ускорение падает от 0 до -a)
# V(t) = v - j*(t - t4)²/2 (скорость уменьшается)
# D(t) = D(t4) + v*(t - t4) - j*(t - t4)³/6
# D(t4) = d (полное расстояние)

J5 = -j
print("J(t):")
pprint(J5)
print()

A5 = integrate(J5, (t, t4, t))
print("A(t) = ∫J(t)dt (с учетом A(t4) = 0) =")
pprint(A5)
print()

V5 = v + integrate(A5, (t, t4, t))
print("V(t) = v + ∫A(t)dt =")
pprint(V5)
print()

D5 = d + integrate(V5, (t, t4, t))
print("D(t) = d + ∫V(t)dt =")
pprint(D5)
print()

# Проверка в момент t5
A5_end = A5.subs(t, t5)
V5_end = V5.subs(t, t5)

print(f"Проверка в момент t5 = {t5}:")
print(f"J(t5) = {J5}")
print(f"A(t5) = {A5_end} = -a")
print(f"Проверка: {simplify(A5_end + a)} = 0")
print(f"V(t5) = {V5_end}")
print(f"V(t5) должно быть = v - a²/(2j)")
print(f"Проверка: {simplify(V5_end - (v - a**2/(2*j)))} = 0")

print("\n" + "="*60 + "\n")

print("Значения в конце интервала (t = t5):")
print(f"J(t5) = {J5}")
print(f"A(t5) = {A5_end}")
print(f"V(t5) = {V5_end}")
```

```python
# @title Этап 8: Интервал t5 ≤ t ≤ t6 (постоянное замедление -a, jerk = 0)
# Запустите эту ячейку после Этапа 7

print("=== ИНТЕРВАЛ 6: t5 ≤ t ≤ t6 ===\n")
print("(постоянное замедление -a, jerk = 0)")

# Границы интервала
t1 = a/j
t2 = v/a
t3 = v/a + a/j
t4 = d/v
t5 = t4 + a/j
t6 = d/v + v/a
t7 = d/v + a/j + v/a

print(f"Границы интервала:")
print(f"t5 = {t5}")
print(f"t6 = {t6}\n")

# V(t5) из предыдущего интервала
V_t5 = v - a**2/(2*j)
print(f"V(t5) = {V_t5}")
print(f"V(t6) = 0 (скорость становится нулевой)\n")

# В этом интервале:
# J(t) = 0 (рынок равен нулю)
# A(t) = -a (постоянное замедление)
# V(t) = V(t5) - a*(t - t5) (скорость линейно уменьшается до 0)
# D(t) = D(t5) + V(t5)*(t - t5) - a*(t - t5)²/2

J6 = S(0)
print("J(t):")
pprint(J6)
print()

A6 = -a
print("A(t):")
pprint(A6)
print()

V6 = V_t5 + integrate(A6, (t, t5, t))
print("V(t) = V(t5) + ∫A(t)dt =")
pprint(V6)
print()

# D(t5) из предыдущего интервала
D5_end = d + v*(t5 - t4) - j*(t5 - t4)**3/6
D5_end = simplify(D5_end)

D6 = D5_end + integrate(V6, (t, t5, t))
print("D(t) = D(t5) + ∫V(t)dt =")
pprint(D6)
print()

# Проверка в момент t6
V6_end = V6.subs(t, t6)
D6_end = D6.subs(t, t6)

print(f"Проверка в момент t6 = {t6}:")
print(f"J(t6) = {J6}")
print(f"A(t6) = {A6} = -a")
print(f"V(t6) = {V6_end} = 0")
print(f"Проверка: {simplify(V6_end)} = 0")

print("\n" + "="*60 + "\n")

print("Значения в конце интервала (t = t6):")
print(f"J(t6) = {J6}")
print(f"A(t6) = {A6}")
print(f"V(t6) = {V6_end}")
print(f"D(t6) = {D6_end}")
```

```python
# @title Этап 9: Интервал t6 ≤ t ≤ t7 (завершение торможения, ускорение от -a до 0, jerk = j)
# Запустите эту ячейку после Этапа 8

print("=== ИНТЕРВАЛ 7: t6 ≤ t ≤ t7 ===\n")
print("(завершение торможения, ускорение от -a до 0, jerk = j)")

# Границы интервала
t1 = a/j
t2 = v/a
t3 = v/a + a/j
t4 = d/v
t5 = t4 + a/j
t6 = d/v + v/a
t7 = d/v + a/j + v/a

print(f"Границы интервала:")
print(f"t6 = {t6}")
print(f"t7 = {t7}\n")

# В этом интервале:
# J(t) = j (постоянный положительный рывок)
# A(t) = -a + j*(t - t6) (ускорение возрастает от -a до 0)
# V(t) = -a*(t - t6) + j*(t - t6)²/2 (скорость от 0 до 0, проходит через минимум)
# D(t) = D(t6) - a*(t - t6)²/2 + j*(t - t6)³/6

J7 = j
print("J(t):")
pprint(J7)
print()

A7 = -a + integrate(J7, (t, t6, t))
print("A(t) = -a + ∫J(t)dt =")
pprint(A7)
print()

V7 = integrate(A7, (t, t6, t))
print("V(t) = ∫A(t)dt (с учетом V(t6) = 0) =")
pprint(V7)
print()

# D(t6) из предыдущего интервала
V_t5 = v - a**2/(2*j)
D5_end = d + v*(t5 - t4) - j*(t5 - t4)**3/6
D6_end = D5_end + V_t5*(t6 - t5) - a*(t6 - t5)**2/2
D6_end = simplify(D6_end)

D7 = D6_end + integrate(V7, (t, t6, t))
print("D(t) = D(t6) + ∫V(t)dt =")
pprint(D7)
print()

# Проверка в момент t7
A7_end = A7.subs(t, t7)
V7_end = V7.subs(t, t7)
D7_end = D7.subs(t, t7)

print(f"Проверка в момент t7 = {t7}:")
print(f"J(t7) = {J7}")
print(f"A(t7) = {A7_end} = 0")
print(f"Проверка: {simplify(A7_end)} = 0")
print(f"V(t7) = {V7_end} = 0")
print(f"Проверка: {simplify(V7_end)} = 0")
print(f"D(t7) = {D7_end} = d")
print(f"Проверка: {simplify(D7_end - d)} = 0")

print("\n" + "="*60 + "\n")

print("Значения в конце интервала (t = t7):")
print(f"J(t7) = {J7}")
print(f"A(t7) = {A7_end}")
print(f"V(t7) = {V7_end}")
print(f"D(t7) = {D7_end}")

print("\n=== ВСЕ ИНТЕРВАЛЫ ЗАВЕРШЕНЫ ===")
print("Условие A: лифт достигает полной скорости")
print(f"Общее время движения: T = t7 = {t7}")
print(f"Формула из файла: T = d/v + a/j + v/a")
print(f"Проверка: {simplify(t7 - (d/v + a/j + v/a))} = 0")

print("\n=== ВСЕ ВРЕМЕННЫЕ ТОЧКИ ДЛЯ УСЛОВИЯ A ===")
print(f"t1 = {t1} (достижение ускорения a)")
print(f"t2 = {t2} (начало уменьшения ускорения)")
print(f"t3 = {t3} (достижение скорости v, ускорение=0)")
print(f"t4 = {t4} = d/v (начало торможения)")
print(f"t5 = {t5} = d/v + a/j (достижение ускорения -a)")
print(f"t6 = {t6} = d/v + v/a (начало завершения торможения)")
print(f"t7 = {t7} = d/v + a/j + v/a (конец движения)")
```
